```python
models = ['frhs']#, 'cgal', 'frhs']#, 'bdck'] # 
# models = ['cgal']

cosmo = {}
cosmo_data = {}


for model in models:
    print('current:', model)
    cosmo[model] = Class()
    cosmo[model].set(pars[model])
    cosmo[model].compute()

    cosmo_data[model] = {'bk': cosmo[model].get_background(), 'pt': cosmo[model].get_perturbations()['scalar']}
```

```python
def plot_bk_de(models):
    fig, ax = plt.subplots()

    # Define a set of line styles to cycle through
    linestyles = ['-', '--', '-.', ':', ]
    de_tag = ['(.)rho_smg', '(.)rho_lambda', '(.)rho_fld']
    for i, model in enumerate(models):
        model_a = 1/(1+cosmo_data[model]['bk']['z'])
        style = linestyles[i % len(linestyles)]  # cycle if more models than styles
        if model=='lcdm':
            rho_de = cosmo_data[model]['bk']['(.)rho_lambda']
        elif model=='w0wa':
            rho_de = cosmo_data[model]['bk']['(.)rho_fld']
        else: 
            rho_de = cosmo_data[model]['bk']['(.)rho_smg']
        ax.semilogx(
            model_a,
            rho_de,
            style,
            label=model,
            alpha=0.7
        )

    ax.set_ylabel(r'$\bar{\rho}_{\text{DE}}$')
    ax.set_xlabel('Scale factor $a$')
    ax.legend()
    return fig, ax

def plot_pt(perturbation, k_idx, models):
    fig, ax = plt.subplots()

    # Define a set of line styles to cycle through
    linestyles = ['-', '--', '-.', ':']
    
    for i, model in enumerate(models):
        style = linestyles[i % len(linestyles)]  # cycle if more models than styles
        ax.semilogx(
            cosmo_data[model]['pt'][k_idx]['a'],
            cosmo_data[model]['pt'][k_idx][perturbation],
            style,
            label=model,
            alpha=0.7
        )
    if perturbation == 'shear_smg':
        ax.set_ylabel(r'$\sigma_{\text{DE}}$')
    elif perturbation == 'delta_smg':
        ax.set_ylabel(r'$\delta_{\text{DE}}$')
    elif perturbation == 'theta_smg':
        ax.set_ylabel(r'$\theta_{\text{DE}}$')
    else:
        ax.set_ylabel(perturbation)

    ax.set_xlabel('Scale factor $a$')
    ax.set_title('k='+str(ks[k_idx]))
    ax.legend()
    return fig, ax
```
